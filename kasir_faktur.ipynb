```python
# kasir_faktur.py

LEBAR = 36

# --- Input ---
nama = input("Nama Pelanggan          : ")
total_awal = float(input("Total Belanja (Rp)      : "))
persen_diskon = float(input("Persentase Diskon (%)   : "))

# --- Perhitungan ---
nominal_diskon = total_awal * persen_diskon / 100
total_bayar = total_awal - nominal_diskon

# --- Format nominal: pemisah ribuan koma, 2 desimal ---
rp_awal = f"Rp{total_awal:,.2f}"
rp_diskon = f"Rp{nominal_diskon:,.2f}"
rp_bayar = f"Rp{total_bayar:,.2f}"

label_diskon = f"Diskon ({persen_diskon:g}%)"

# --- Cetak nota ---
print()
print("=" * LEBAR)
print(f"{'NOTA PEMBAYARAN':^{LEBAR}}")
print(f"{nama:^{LEBAR}}")
print("=" * LEBAR)
print(f"{'Total Awal':<16}{rp_awal:>20}")
print(f"{label_diskon:<16}{rp_diskon:>20}")
print("-" * LEBAR)
print(f"{'Total Bayar':<16}{rp_bayar:>20}")
print("=" * LEBAR)
```
